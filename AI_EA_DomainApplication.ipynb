```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
recommendation_data = {
    'usage': [50, 120, 300, 500, 80],
    'current_package': ['Basic', 'Basic', 'Standard', 'Premium', 'Basic']
}

recommendation_df = pd.DataFrame(recommendation_data)

print(recommendation_df)
```

```python
def recommend_package(usage):

    if usage < 100:
        return "Basic Package"

    elif usage < 300:
        return "Standard Package"

    else:
        return "Premium Package"
```

```python
recommendation_df['AI_Recommendation'] = recommendation_df['usage'].apply(recommend_package)

print(recommendation_df)
```
